# PROMPTS

```python
from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder
```

```python
history = []
role = "physicist"
prompt = ChatPromptTemplate.from_messages([
    ('system', f'Answer the question as a {role}'),
    MessagesPlaceholder("history"),
    ('human', '{question}'),
])
```

```python
from langchain.chat_models import init_chat_model
model = init_chat_model(
    model="qwen3.5:9b",
    model_provider="ollama",
    temperature=.6,
    validate_model_on_init=True,
    reasoning=False
)
```

```python
chain = prompt | model
```

```python
from langchain.messages import HumanMessage, AIMessage
```

```python
history = []
total_response = ""

for chunk in chain.stream({
    "question": "What is the speed of light?",
    "history": history
}):
    print(chunk.content, end="")
    total_response += chunk.content

history.append(HumanMessage(content="What is the speed of light?"))
history.append(AIMessage(content=total_response))
```

```python
for chunk in chain.stream({
    "question": "What was the last question and how did you answer it in summary?",
    "history": history
}):
    print(chunk.content, end="")
    total_response += chunk.content
```

### few-shot samples

```python
history = []
total_response = ""

question = 'What are the benefits of RAG systems?'

prompt = ChatPromptTemplate.from_messages([
    ('system', 'Answer the question as a ML engineer'),
    ('human', 'What how to build a LLM?'),
    ('ai', 'Lets answer for Efee.. You can build a LLM by using a transformer model and a dataset of text. You can use a library like Hugging Face to get the data and train the model.'),
    ('human', 'What is the best way to train a LLM?'),
    ('ai', 'Lets answer for Efee..You can train a LLM by using a dataset of text. You can use a library like Hugging Face to get the data and train the model.'),
    MessagesPlaceholder("history"),
    ('human', '{question}'),
])

chain = prompt | model

for chunk in chain.stream(
    {'question': question,
    'history': history
}):

    print(chunk.content, end="")
    total_response += chunk.content

history.append(HumanMessage(content=question))
history.append(AIMessage(content=total_response))
```

```python
question = "what was the last question?"

for chunk in chain.stream(
    {'question': question,
    'history': history
}):

    print(chunk.content, end="")
    total_response += chunk.content

history.append(HumanMessage(content=question))
history.append(AIMessage(content=total_response))
```
