# CHAT MODELS

```python
from langchain.chat_models import init_chat_model
from langchain.messages import HumanMessage
```

```python
model = init_chat_model(
    model_provider='ollama',
    model='qwen3.5:9b',
    validate_model_on_init=True,
    temperature=0.6)

message = [HumanMessage("How long it took to end world war 2?")]
```

```python
response= model.invoke(message)
```

```python
print("METADATA: \n\n", response.usage_metadata, "\n\nCONTENT: \n\n", response.content)
```

```python
for chunk in model.stream(message):
    print(chunk.content, end="")
```

```python
message1 = [HumanMessage("How long it took to end world war 1?")]
message2 = [HumanMessage("How long it took to end world war afghanistan - usa?")]

for response in model.batch([message1, message2]):
    print(response.content, end="\n---------------------------------------\n")
```

## batch optimization for local GPU

```python
config = {"max_concurrency": 2}
responses = model.batch([message1, message2], config=config)

for i, res in enumerate(responses, 1):
    print(f"\n[Answer {i}]:")
    print(res.content)
```

```python
await model.ainvoke(message)
print("While running model inference, still print can be used here")
```

```python
messages = [HumanMessage(content="Explain quantum physics in basics.")]

async for chunk in model.astream(messages):
    print(chunk.content, end="", flush=True)
```

```python
q1 = [HumanMessage(content="Fastest planet in solar system?")]
q2 = [HumanMessage(content="How fast is the light?")]

responses = await model.abatch([q1, q2], config={"max_concurrency": 2})

for r in responses:
    print(r.content)
    print("-" * 20)
```
