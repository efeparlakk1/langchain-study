# STRUCTURED OUTPUTS

```python
from dotenv import load_dotenv
import os
load_dotenv()
MODEL_PROVIDER=os.getenv("MODEL_PROVIDER")
MODEL=os.getenv("MODEL")
```

```python
from pydantic import BaseModel, Field
from langchain.chat_models import init_chat_model
```

```python
class Movie(BaseModel):
    title: str = Field(description="Name of the movie")
    year: int = Field(description="Year of the movie")

model=init_chat_model(
    model_provider=MODEL_PROVIDER,
    model=MODEL,
    reasoning=False,
    validate_model_on_init=False,
    temperature=0)
```

```python
structured = model.with_structured_output(Movie, method="function_calling")
movie = structured.invoke("Give me a information about Inception")
print("TITLE: ",movie.title)
print("YEAR: ",movie.year)
#include_raw=True for debug in with_structured_output
```
