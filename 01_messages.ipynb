# MESSAGES

```python
from langchain.messages import SystemMessage, HumanMessage, AIMessage, ToolMessage
```

```python
system_message = SystemMessage(content="You are a helpful assistant that can answer questions and help with tasks, but all answers will start with 'Hey Efe!'")
human_message = HumanMessage(content="What is the weather like in Antalya?")
#ai_message = AIMessage(content="Hey Efe! The capital of France is Paris.")

print("SYSTEM_MES: ", system_message.type, system_message.content)
print("HUMAN_MES: ", human_message.type, human_message.content)
#print("AI_MES: ", ai_message.type, ai_message.content)

messages = [system_message, human_message]
```

```python
from langchain.chat_models import init_chat_model
model=init_chat_model(
    model_provider = 'ollama',
    model='qwen3.5:9b',
    validate_model_on_init=True,
    temperature=0.6
)
```

```python
model.invoke(messages).content_blocks[0]['type']
```

```python
ai_message = AIMessage(
    content="Hey Efe! Checking the weather...", 
    tool_calls=[
        {
            "name": "get_weather",
            "args": {"location": "Antalya"},
            "id": "call_1",
            "type": "tool_call"
        }
    ]
)

tool_message = ToolMessage(
    content="72F", 
    tool_call_id="call_1"
)
```

```python
messages = [system_message, human_message, ai_message, tool_message]
```

```python
model.invoke(messages).content_blocks
```
