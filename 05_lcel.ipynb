# RUNNABLE AND LCEL

```python
from langchain.chat_models import init_chat_model
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
```

```python
from dotenv import load_dotenv
import os
load_dotenv()
MODEL = os.getenv("MODEL")
MODEL_PROVIDER= os.getenv("MODEL_PROVIDER")
```

```python
model = init_chat_model(model=MODEL, model_provider=MODEL_PROVIDER, temperature=0, reasoning=False)
```

```python
prompt= ChatPromptTemplate.from_messages([
    ('system', 'You are a helpful assistant.'),
    ('human', '{topic}'),
])
```

```python
parser = StrOutputParser()
```

```python
step1 = prompt.invoke({"topic": "LCEL"})
print(type(step1))          # ChatPromptValue veya mesaj listesi
step2 = model.invoke(step1)
print(type(step2))          # AIMessage
step3 = parser.invoke(step2)
print(type(step3))          # str
```

```python
chain = prompt | model | parser
```

```python
chain.invoke({"topic": "LCEL"})
```

```python
chain.get_graph().print_ascii()
```

```python
with_parser = prompt | model | parser
without_parser = prompt | model

result1 = with_parser.invoke({'topic': 'LCEL'})
result2 = without_parser.invoke({'topic': 'LCEL'})

print("WITH PARSER: ", result1)
print("/nWITHOUT PARSER: ", result2)
```

```python
print(type(result1))    # AIMessage
print(type(result2))  # str
```
